# Example Analysis Queries

This document contains example queries that show how to use the Gold tables for common risk-analysis use cases. 

```sql
SELECT
  zone_id,
  event_dt,
  total_risk_score,
  event_count
FROM main.gold.daily_risk_zone
WHERE event_dt >= DATEADD(day, -1, CURRENT_DATE())
ORDER BY total_risk_score DESC
LIMIT 10;
```

```sql
WITH target AS (
  SELECT h3_longlatash3string(-0.12, 51.5, 8) AS target_h3
)
SELECT
  event_dt,
  total_risk_score,
  event_count
FROM main.gold.daily_risk_cell
CROSS JOIN target
WHERE h3_ix = target.target_h3
ORDER BY event_dt;
```

```sql
SELECT
  h3_ix,
  SUM(total_risk_score) AS total_risk_score,
  SUM(event_count)      AS event_count,
  AVG(avg_risk_score)   AS avg_risk_score
FROM main.gold.daily_risk_cell
WHERE event_dt BETWEEN DATE '2026-05-01' AND DATE '2026-05-07'
GROUP BY h3_ix
HAVING SUM(event_count) >= 10
ORDER BY total_risk_score DESC;
```

```sql
SELECT
  event_dt,
  COUNT(DISTINCT h3_ix) AS active_cells,
  SUM(event_count)      AS total_events
FROM main.gold.daily_risk_cell
GROUP BY event_dt
ORDER BY event_dt;
```

```sql
SELECT
  event_dt,
  total_risk_score,
  event_count,
  avg_cell_risk_score
FROM main.gold.daily_risk_zone
WHERE zone_id = '<some_h3_zone_id>'
ORDER BY event_dt;
```
